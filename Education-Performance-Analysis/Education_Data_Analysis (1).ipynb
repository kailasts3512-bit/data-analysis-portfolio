```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("education_student_performance.csv")

df.head()
```

```python
df.info()
```

```python
df.isnull().sum()
```

```python
df.describe()
```

```python
course_score = df.groupby("Course")["Final_Score"].mean()

course_score
```

```python
course_score.plot(kind="bar")
plt.title("Average Score by Course")
plt.xlabel("Course")
plt.ylabel("Average Score")
plt.show()
```

```python
plt.scatter(df["Study_Hours_per_Week"], df["Final_Score"])
plt.title("Study Hours vs Final Score")
plt.xlabel("Study Hours per Week")
plt.ylabel("Final Score")
plt.show()
```

```python
result = df["Result"].value_counts()

result.plot(kind="pie", autopct='%1.1f%%')
plt.title("Pass vs Fail Distribution")
plt.show()
```
